```python
# solution 12:
num = int(input("Enter a number to check if it is prime: "))

if num > 1:
    is_prime = True 
    
    for i in range(2, num):
        if num % i == 0:
            is_prime = False 
            break
            
    if is_prime:
        print(f"{num} is a Prime number.")
    else:
        print(f"{num} is not a Prime number.")
else:
    print(f"{num} is not a Prime number.")
```

```python
# solution 16:
num = int(input("Enter a number: "))

factorial = 1

if num < 0:
    print("Factorial does not exist for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1")
else:
    for i in range(1, num + 1):
        factorial = factorial * i
    print(f"The factorial of {num} is {factorial}")
```

```python
# solution 20:
import random

secret_number = random.randint(1, 20)

chances = 3
current_chance = 0

print("🎮 Number Guessing Game! 🎮")
print("Gues Number between 1-20, you have 3 chances ony.")
print("-" * 55)

while current_chance < chances:
    user_guess = int(input(f"Attempt {current_chance + 1}: Enter your number: "))
    
    current_chance += 1
    
    if user_guess == secret_number:
        print("Congrats you guess right number!")
        break
    else:
        if current_chance == chances:
            print(f"\nBetter luck next time, the correct number is: {secret_number}")
        elif user_guess < secret_number:
            print("Wrong, Guess big number.")
        else:
            print("Wrong, Guess small number.")
```
